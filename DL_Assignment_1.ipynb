STEP 1 - Import mnist (number sample dataset)

```python
from tensorflow.keras.datasets import mnist
(train_images, train_labels), (test_images, test_labels) = mnist.load_data()
```

STEP 2 - Initializing neural network layers

```python
from tensorflow import keras
from tensorflow.keras import layers
model = keras.Sequential()
model.add(layers.Dense(512, activation='relu', input_shape=(28 * 28,)))
model.add(layers.Dense(10, activation='softmax'))
```

STEP 3 - Initializing optimizer & loss function

```python
model.compile(optimizer="rmsprop",
              loss="categorical_crossentropy",
              metrics=["accuracy"])
```

STEP 4 - Prepare, train, and test the images

```python
train_images = train_images.reshape((60000, 28 * 28))
train_images = train_images.astype("float32") / 255
test_images = test_images.reshape((10000, 28 * 28))
test_images = test_images.astype("float32") / 255
```

STEP 5 - Convert integer targets into categorical targets

```python
from tensorflow.keras.utils import to_categorical
train_labels = to_categorical(train_labels)
test_labels = to_categorical(test_labels)
```

STEP 6 - Using 'fit' to train the network

```python
model.fit(train_images, train_labels, epochs=5, batch_size=128)
```

STEP 7 - Use evaluate to test the network and print out the test accuracy

```python
test_loss, test_acc = model.evaluate(test_images, test_labels)
print(f"test_acc: {test_acc}")
```

Test Accuracy: 0.9805999994277954
